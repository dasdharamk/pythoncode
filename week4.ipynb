```python
#adding an alternative
sales = int(input("Enter sales>>"))
if sales >= 500:
    print ("Sales are GOOD.")
else:
    print("Sales are less than 500 or sales are not GOOD")
print("End Processing.")
```

```python
#multiple input ranges and condition order
sales = int(input("Enter sales>>")) 
if sales >= 500:
    print ("Sales are GOOD.") 
elif sales >=250:
    print("Sales are 250 or more but less than 500.")
else:
    print("Sales are less than 250.")
print ("End processing.")
```
