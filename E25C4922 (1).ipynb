```python
# Class Trip Budget
destination = input("Where is your Destination?: ")
participants = int(input("Participants: "))
transportation = int(input("Transportation cost: "))
admission_per_person = int(input("Admission per person: "))
meal_per_person = int(input("Meal cost per person: "))
budget = int(input("Available budget: "))

# Calculate costs
admission_total = participants * admission_per_person
meals_total = participants * meal_per_person
total_trip_cost = transportation + admission_total + meals_total
average_cost = total_trip_cost / participants
budget_remaining = budget - total_trip_cost
within_budget = total_trip_cost <= budget

# Display summary
print("\nCLASS TRIP BUDGET SUMMARY")
print("Destination:", destination)
print("Participants:", participants)
print("Transportation:", transportation, "yen")
print("Admission total:", admission_total, "yen")
print("Meals total:", meals_total, "yen")
print("Total trip cost:", total_trip_cost, "yen")
print("Average cost per participant:", average_cost, "yen")
print("Available budget:", budget, "yen")
print("Budget remaining:", budget_remaining, "yen")
print("Within budget:", within_budget)
```
